```python
RESULTS_DIR = "results"
LR = "0.1"
WD = "0.0"
SEEDS = [0]

MUON_VARIANTS = [
    ("polarexpress", 5),
    ("ns5", 5),
    ("jordan5", 5),
    ("svd_ref", 1),
    ("you6", 6),
]

SAVE_DIR = "figures"
SAVE_PREFIX = f"cifar10_lr{LR}_wd{WD}_seeds{'-'.join(map(str,SEEDS))}"

import os, io, pickle
from typing import Any, Dict, List, Tuple
import numpy as np
import torch
import matplotlib as mpl
import matplotlib.pyplot as plt

mpl.rcParams["text.usetex"] = False
mpl.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 300,
    "font.size": 10,
    "axes.titlesize": 11,
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 9,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.linewidth": 0.8,
})

Results = Dict[str, Any]

def torch_load_cpu(path: str):
    return torch.load(path, map_location="cpu", weights_only=False)

def pickle_load(path: str):
    with open(path, "rb") as f:
        return pickle.load(f)

def load_results(path: str) -> Results:
    try:
        obj = torch_load_cpu(path)
    except Exception:
        try:
            with open(path, "rb") as f:
                buf = io.BytesIO(f.read())
            obj = torch.load(buf, map_location="cpu", weights_only=False)
        except Exception:
            obj = pickle_load(path)
    if not isinstance(obj, dict):
        raise TypeError(f"{path} does not contain a dict.")
    return obj

def to_cpu_tensor(x) -> torch.Tensor:
    if isinstance(x, torch.Tensor):
        return x.detach().cpu()
    if isinstance(x, (list, tuple, np.ndarray)):
        return torch.tensor(x)
    return torch.tensor([x])

def mean_std_stack(arrs: List[np.ndarray]) -> Tuple[np.ndarray, np.ndarray]:
    A = np.stack(arrs, axis=0)
    return A.mean(axis=0), A.std(axis=0, ddof=0)

def spectral_norms_from_singular_values(svals_list) -> np.ndarray:
    out = []
    for svals in svals_list:
        t = to_cpu_tensor(svals).float().flatten()
        out.append(float(t.max().item()) if t.numel() else np.nan)
    return np.array(out, dtype=float)

def fname_adam(seed: int) -> str:
    return f"update-adam-lr-{LR}-wd-{WD}-seed-{seed}.pkl"

def fname_hyp(seed: int) -> str:
    return f"update-hyperspherical_descent-lr-{LR}-wd-{WD}-seed-{seed}.pkl"

def fname_muon(msign: str, msteps: int, seed: int) -> str:
    return f"update-manifold_muon-msign-{msign}-msignsteps-{msteps}-lr-{LR}-wd-{WD}-seed-{seed}.pkl"

def label_muon(msign: str, msteps: int) -> str:
    return f"manifold_muon/{msign})"

expected: List[str] = []
expected.append("adam")
expected.append("hyperspherical_descent")
for msign, msteps in MUON_VARIANTS:
    expected.append(label_muon(msign, msteps))

runs: Dict[str, List[Results]] = {lab: [] for lab in expected}
missing = []

for seed in SEEDS:
    p = os.path.join(RESULTS_DIR, fname_adam(seed))
    if os.path.isfile(p):
        runs["adam"].append(load_results(p))
    else:
        missing.append(p)

    p = os.path.join(RESULTS_DIR, fname_hyp(seed))
    if os.path.isfile(p):
        runs["hyperspherical_descent"].append(load_results(p))
    else:
        missing.append(p)

    for msign, msteps in MUON_VARIANTS:
        lab = label_muon(msign, msteps)
        p = os.path.join(RESULTS_DIR, fname_muon(msign, msteps, seed))
        if os.path.isfile(p):
            runs[lab].append(load_results(p))
        else:
            missing.append(p)

present = {k: v for k, v in runs.items() if len(v) > 0}
if not present:
    raise SystemExit("No result files found for the given SETTINGS.")
if missing:
    print("Missing files (ignored):")
    for p in missing:
        print(" ", p)

agg = {}
for lab, lst in present.items():
    epochs_declared = int(lst[0].get("epochs", 0))
    losses = [np.array(r["epoch_losses"], dtype=float) for r in lst]
    times  = [np.array(r["epoch_times"], dtype=float)  for r in lst]

    if len(losses) > 1:
        loss_m, loss_s = mean_std_stack(losses)
        time_m, time_s = mean_std_stack(times)
    else:
        loss_m, loss_s = losses[0], np.zeros_like(losses[0])
        time_m, time_s = times[0],  np.zeros_like(times[0])

    test_accs  = np.array([float(r["test_acc"])  for r in lst], dtype=float)
    train_accs = np.array([float(r["train_acc"]) for r in lst], dtype=float)

    norms_per_seed = []
    for r in lst:
        norms = []
        for n in r["norms"]:
            t = to_cpu_tensor(n).float().flatten()
            norms.append(float(t.item()) if t.numel() == 1 else float(t.mean().item()))
        norms_per_seed.append(np.array(norms, dtype=float))
    norms_per_seed = np.stack(norms_per_seed, axis=0)
    norms_m = norms_per_seed.mean(axis=0)
    norms_s = norms_per_seed.std(axis=0, ddof=0)

    spec_per_seed = []
    for r in lst:
        spec_per_seed.append(spectral_norms_from_singular_values(r["singular_values"]))
    spec_per_seed = np.stack(spec_per_seed, axis=0)
    spec_m = np.nanmean(spec_per_seed, axis=0)
    spec_s = np.nanstd(spec_per_seed, axis=0, ddof=0)

    agg[lab] = {
        "epochs_declared": epochs_declared,
        "loss_m": loss_m, "loss_s": loss_s,
        "time_m": time_m, "time_s": time_s,
        "test_m": float(test_accs.mean()),  "test_s": float(test_accs.std(ddof=0)),
        "train_m": float(train_accs.mean()), "train_s": float(train_accs.std(ddof=0)),
        "norms_m": norms_m, "norms_s": norms_s,
        "spec_m": spec_m, "spec_s": spec_s,
    }

labels = list(agg.keys())
n_layers = len(next(iter(agg.values()))["norms_m"])
layer_idx = np.arange(n_layers)
layer_labels = [f"W{i+1}" for i in range(n_layers)]

os.makedirs(SAVE_DIR, exist_ok=True)

plt.figure(figsize=(7.2, 4.2))
for lab in labels:
    m = agg[lab]["loss_m"]; s = agg[lab]["loss_s"]
    x_epochs = np.arange(1, len(m) + 1)
    plt.plot(x_epochs, m, label=lab)
    plt.fill_between(x_epochs, m - s, m + s, alpha=0.2)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title(f"Loss vs Epoch")
plt.legend(loc="best")
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_loss.png"), bbox_inches="tight")
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_loss.pdf"), bbox_inches="tight")
plt.show()

plt.figure(figsize=(8.0, 4.2))
x = np.arange(len(labels))
width = 0.38
test_means  = [agg[lab]["test_m"]  for lab in labels]
test_stds   = [agg[lab]["test_s"]  for lab in labels]
train_means = [agg[lab]["train_m"] for lab in labels]
train_stds  = [agg[lab]["train_s"] for lab in labels]
plt.bar(x - width/2, test_means,  width, yerr=test_stds,  capsize=3, label="Test")
plt.bar(x + width/2, train_means, width, yerr=train_stds, capsize=3, label="Train")
plt.xticks(x, labels, rotation=18, ha="right")
plt.ylabel("Accuracy (%)")
plt.title(f"Final Accuracy  ")
plt.legend(loc="best")
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_acc.png"), bbox_inches="tight")
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_acc.pdf"), bbox_inches="tight")
plt.show()

plt.figure(figsize=(8.0, 4.0))
avg_time_means = [float(np.mean(agg[lab]["time_m"])) for lab in labels]
avg_time_stds  = [float(np.mean(agg[lab]["time_s"])) for lab in labels]
plt.bar(x, avg_time_means, yerr=avg_time_stds, capsize=3)
plt.xticks(x, labels, rotation=18, ha="right")
plt.ylabel("sec/epoch")
plt.title(f"Average Epoch Time ")
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_time.png"), bbox_inches="tight")
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_time.pdf"), bbox_inches="tight")
plt.show()

plt.figure(figsize=(8.6, 4.2))
group_width = 0.8
bar_w = group_width / max(1, len(labels))
for i, lab in enumerate(labels):
    offs = layer_idx - group_width/2 + bar_w*(i + 0.5)
    plt.bar(offs, agg[lab]["norms_m"], width=bar_w, yerr=agg[lab]["norms_s"], capsize=2, label=lab)
plt.xticks(layer_idx, layer_labels)
plt.ylabel("||W||_F")
plt.title(f"Frobenius Norms per Layer ")
plt.legend(loc="best")
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_fro.png"), bbox_inches="tight")
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_fro.pdf"), bbox_inches="tight")
plt.show()

plt.figure(figsize=(8.6, 4.2))
for lab in labels:
    plt.errorbar(layer_idx, agg[lab]["spec_m"], yerr=agg[lab]["spec_s"], marker="o", capsize=2, label=lab)
plt.xticks(layer_idx, layer_labels)
plt.ylabel("||W||_2 (max σ)")
plt.title(f"Spectral Norms per Layer ")
plt.legend(loc="best")
plt.tight_layout()
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_spec.png"), bbox_inches="tight")
plt.savefig(os.path.join(SAVE_DIR, f"{SAVE_PREFIX}_spec.pdf"), bbox_inches="tight")
plt.show()
```
